# Live Financial Market Analytics & Forecasting

This notebook collects financial market data from an external API
and prepares it for further analysis and forecasting.

```python
import requests
import pandas as pd
import os

from dotenv import load_dotenv
```

```python
load_dotenv()

API_KEY = os.getenv("ALPHA_VANTAGE_API_KEY")

print("API Key Loaded:", API_KEY is not None)
```

```python
SYMBOL = "AAPL"

print("Selected Stock:", SYMBOL)
```

```python
BASE_URL = "https://www.alphavantage.co/query"

params = {
    "function": "TIME_SERIES_DAILY",
    "symbol": SYMBOL,
    "outputsize": "compact",
    "apikey": API_KEY
}

print("API request parameters created successfully.")
```

```python
response = requests.get(BASE_URL, params=params)

print("Status Code:", response.status_code)
```

```python
data = response.json()

print(data.keys())
```

```python
time_series = data["Time Series (Daily)"]

print(type(time_series))
print(list(time_series.items())[:2])
```

```python
df = pd.DataFrame.from_dict(time_series, orient="index")

df.head()
```

```python
df.columns = [
    "Open",
    "High",
    "Low",
    "Close",
    "Volume"
]

df.head()
```

```python
df.index = pd.to_datetime(df.index)

df.index.name = "Date"

df.head()
```

```python
numeric_columns = ["Open", "High", "Low", "Close", "Volume"]

df[numeric_columns] = df[numeric_columns].astype(float)

df.info()
```

```python
df = df.sort_index()

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv("data/market_data.csv", index=True)

print("Market data saved successfully.")
```
